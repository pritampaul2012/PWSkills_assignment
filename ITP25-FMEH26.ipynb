**Q.10** Write a function that accepts marks of five subjects
and returns
● Total
● Percentage
● Grade

```python
def evaluate_marks(marks):
    
    total = sum(marks)
    percentage = (total / 500) * 100

    if percentage >= 90:
        grade = "A+"
    elif percentage >= 80:
        grade = "A"
    elif percentage >= 70:
        grade = "B"
    elif percentage >= 60:
        grade = "C"
    elif percentage >= 50:
        grade = "D"
    else:
        grade = "F"

    return total, percentage, grade


# Collect marks for 5 subjects from the user
student_marks = []
subjects = ["Subject 1", "Subject 2", "Subject 3", "Subject 4", "Subject 5"]

print("Enter marks out of 100 for each subject:")
for subject in subjects:
    while True:
        try:
            score = float(input(f"Enter marks for {subject}: "))
            if 0 <= score <= 100:
                student_marks.append(score)
                break
            else:
                print("Marks must be between 0 and 100. Try again.")
        except ValueError:
            print("Invalid input. Please enter a valid numerical value.")

# Calculate results using the function
total, pct, grade = evaluate_marks(student_marks)

# Display results
print(f"Marks Entered: {student_marks}")
print(f"Total Marks:   {total:.2f} / 500")
print(f"Percentage:    {pct:.2f}%")
print(f"Grade:         {grade}")
```

**Q.11** Given
employees = {
"Amit":45000,
"Neha":55000,
"Riya":70000,
"Raj":65000
}
Write a program to
● Increase every salary by 10%
● Find highest-paid employee
● Find average salary

```python
employees = {
    "Amit": 45000,
    "Neha": 55000,
    "Riya": 70000,
    "Raj": 65000
}

# 1. Increase every salary by 10%
updated_employees = {}
for name in employees:
    current_salary = employees.get(name, 0)
    updated_employees[name] = round(current_salary * 1.10, 2)

print("Updated Salaries (+10%):", updated_employees)

# 2. Find highest-paid employee using
highest_paid_employee = max(updated_employees, key=updated_employees.get)
highest_salary = updated_employees.get(highest_paid_employee)
print(f"Highest-paid employee: {highest_paid_employee} ({highest_salary})")

# 3. Find average salary using
avg_salary = sum(updated_employees.values()) / len(updated_employees)
print(f"Average salary: {avg_salary:.2f}")
```

**Q.12** Given
students = {
"Rahul":{"Math":90,"Science":88},
"Aman":{"Math":78,"Science":80},
"Neha":{"Math":95,"Science":92}
}
Display
● Total marks
● Average marks
● Top score

```python
students = {
    "Rahul": {"Math": 90, "Science": 88},
    "Aman": {"Math": 78, "Science": 80},
    "Neha": {"Math": 95, "Science": 92}
}

for name, scores in students.items():
    total_score = sum(scores.values())
    avg_score = total_score / len(scores)
    top_subject_score = max(scores.values())
    
    print(f"Student: {name}")
    print(f"    Total Marks: {total_score}")
    print(f"    Average Marks: {avg_score:.2f}")
    print(f"    Top Score: {top_subject_score}\n")
```

**Q.14** Create reusable functions
● Deposit
● Withdraw
● Check Balance
Handle insufficient balance using exception handling.

```python
# Custom exception for insufficient funds
class InsufficientBalanceError(Exception):
    pass

account_balance = 1000.0

def check_balance():
    print(f"Current Balance: ${account_balance:.2f}")
    return account_balance

def deposit(amount):
    global account_balance
    if amount <= 0:
        print("Deposit amount must be greater than zero.")
        return account_balance
    account_balance += amount
    print(f"Successfully deposited: ${amount:.2f}")
    return account_balance

def withdraw(amount):
    global account_balance
    if amount <= 0:
        print("Withdrawal amount must be greater than zero.")
        return account_balance
    
    if amount > account_balance:
        raise InsufficientBalanceError(
            f"Withdrawal denied: Requested ${amount:.2f}, but available balance is only ${account_balance:.2f}."
        )
    
    account_balance -= amount
    print(f"Successfully withdrew: ${amount:.2f}")
    return account_balance

# Demonstration
check_balance()
deposit(500)
check_balance()

# Test valid withdrawal
try:
    withdraw(300)
    check_balance()
except InsufficientBalanceError as error:
    print(f"Exception: {error}")

# Test invalid withdrawal (trigger exception)
try:
    withdraw(2000)
except InsufficientBalanceError as error:
    print(f"Exception: {error}")
finally:
    check_balance()
```
